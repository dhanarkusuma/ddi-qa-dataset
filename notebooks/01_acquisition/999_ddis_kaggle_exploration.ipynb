## Setup Config

```python
DATASET_DIR = "../../datasets/"
DATASET_REPOSITORY = "mghobashy/drug-drug-interactions"

import pathlib
OUTPUT_DIR = pathlib.Path(DATASET_DIR)
print(f"Dataset Directory: {OUTPUT_DIR}")

FILE_DDI = "db_drug_interactions.csv"
DATASET_PATH = OUTPUT_DIR.joinpath(FILE_DDI)
print(f"Drug-Drug Interactions Dataset Location: {DATASET_PATH}")

FILE_UNIQUE_DRUG_DDI = "unique_drug_interactions.csv"
UNIQUE_DRUG_DDI_PATH = OUTPUT_DIR.joinpath(FILE_UNIQUE_DRUG_DDI)
```

## Downloading Dataset

```python
import kagglehub

kagglehub.dataset_download(DATASET_REPOSITORY, output_dir=str(DATASET_DIR))
```

## Read Drug Drug Interactions Dataset

```python
import pandas as pd

df = pd.read_csv(DATASET_PATH)
print(df.head())
```

```python
print(f"Total row dataset drug-drug interactions: {df.shape[0]}")
```

### Unique Drug 

```python
unique_drug1 = df["Drug 1"].unique()
unique_drug2 = df["Drug 2"].unique()
print(f"Unique Drug 1: {len(unique_drug1)}")
print(f"Unique Drug 2: {len(unique_drug2)}")

unique_drugs = pd.concat([
    df["Drug 1"],
    df["Drug 2"]
]).drop_duplicates()
unique_drugs = unique_drugs.to_frame(name="drug")
print(f"Unique drug in drug-drug interactions: {len(unique_drugs)}")

unique_drugs.to_csv(UNIQUE_DRUG_DDI_PATH, index=False)
print(f"List of unique drug stored at {UNIQUE_DRUG_DDI_PATH}")
```
