```python
PROJECT_DIR = "../.."

import pathlib

MAX_WORKERS = 4                  # simultaneous calls (3/4/5/...)
MAX_REQUESTS_PER_SECOND = 10     # global cap across all workers (RxNav documents 20 req/s per IP)
REQUEST_TIMEOUT = 20             # seconds
MAX_RETRIES = 5                  # on 429 / 5xx / connection errors, with exponential backoff

OUTPUT_DIR = pathlib.Path(PROJECT_DIR)
print(f"Dataset Directory: {OUTPUT_DIR}")

FILE_BRAND_RXNORM = "rxnorm/rxnorm_brands.csv"
RXNORM_BRAND_PATH = OUTPUT_DIR.joinpath(FILE_BRAND_RXNORM)
print(f"RxNorm Brand Path: {RXNORM_BRAND_PATH}")

FILE_BRAND_PRODUCT_RXNORM = "rxnorm/rxnorm_brand_products.csv"
RXNORM_BRAND_PRODUCT_PATH = OUTPUT_DIR.joinpath(FILE_BRAND_PRODUCT_RXNORM)
print(f"RxNorm Brand Path: {RXNORM_BRAND_PRODUCT_PATH}")

FILE_BRAND_RXNORM_ERR = "notebooks/reports/brand_rxnorm_err.csv"
RXNORM_BRAND_ERR_PATH = OUTPUT_DIR.joinpath(FILE_BRAND_RXNORM_ERR)
print(f"RxNorm Brand Path: {RXNORM_BRAND_ERR_PATH}")
```

```python
import threading
import time

class RateLimiter:
    """Thread-safe limiter: hands out evenly spaced time slots to all workers."""
 
    def __init__(self, rate_per_second: float):
        self._interval = 1.0 / rate_per_second
        self._lock = threading.Lock()
        self._next_slot = time.monotonic()
 
    def wait(self) -> None:
        with self._lock:
            now = time.monotonic()
            slot = max(now, self._next_slot)
            self._next_slot = slot + self._interval
        delay = slot - now
        if delay > 0:
            time.sleep(delay)
```

```python
import time
import requests
from typing import List

class RxNormConceptFetcher:
    BASE_URL = "https://rxnav.nlm.nih.gov/REST"

    def get_related(self, rxcui: str, ttys: List[str]):
        try:
            url = f"{self.BASE_URL}/rxcui/{rxcui}/related.json"
    
            response = requests.get(
                url,
                params={"tty": " ".join(ttys)},
                timeout=20
            )
    
            response.raise_for_status()
            data = response.json()
    
            concepts = (
                data.get("relatedGroup", {})
                .get("conceptGroup", [])
            )
    
            if not concepts:
                return None

            results = []
            for item in concepts:
                if item.get("tty") not in ttys:
                    continue
                    
                concept_properties = item.get("conceptProperties", [])
                for concept_property in concept_properties:
                    rx_cui = concept_property.get("rxcui")
                    ingredient_name = concept_property.get("name")
                    
                    results.append({
                        "rx_cui": rx_cui,
                        "name": ingredient_name,
                        "tty": concept_property.get("tty")
                    })
                
            return results
        except Exception as e:
            print(f"[ALLCONCEPTS ERROR] {tty}: {e}")
            return None

    def _get_allconcepts(self, tty: str):
        try:
            url = f"{self.BASE_URL}/allconcepts.json"
    
            response = requests.get(
                url,
                params={"tty": tty},
                timeout=20
            )
    
            response.raise_for_status()
            data = response.json()
    
            concepts = (
                data.get("minConceptGroup", {})
                .get("minConcept", [])
            )
    
            if not concepts:
                return None

            results = []
            for item in concepts:
                if item.get("tty") != tty:
                    continue
                    
                rx_cui = item.get("rxcui")
                ingredient_name = item.get("name")
                
                results.append({
                    "rx_cui": rx_cui,
                    "name": ingredient_name
                })
                
            return results
        except Exception as e:
            print(f"[ALLCONCEPTS ERROR] {tty}: {e}")
            return None

    def get_all_brand(self):
        tty = "BN"
            
        return self._get_allconcepts(tty)
```

```python
from concurrent.futures import ThreadPoolExecutor, as_completed
from typing import Dict, List, Tuple
 
import requests
from requests.adapters import HTTPAdapter
from tqdm import tqdm
from urllib3.util.retry import Retry

class ConcurrentRxNormFetcher(RxNormConceptFetcher):  # noqa: F821  (defined in an earlier cell)
    """Same API as RxNormConceptFetcher.get_related, but safe to call from many threads.
 
    Differences from the original:
      * shared Session with a connection pool sized to the worker count
      * global rate limit + automatic retry/backoff (honours Retry-After)
      * raises on failure instead of returning None, so a network error is never
        confused with "this concept has no related items" (returns [] for that case)
    """
 
    def __init__(self, max_workers: int = MAX_WORKERS, rps: float = MAX_REQUESTS_PER_SECOND):
        self.limiter = RateLimiter(rps)
        self.session = requests.Session()
        retry = Retry(
            total=MAX_RETRIES,
            backoff_factor=1.0,
            status_forcelist=[429, 500, 502, 503, 504],
            allowed_methods=["GET"],
            respect_retry_after_header=True,
        )
        adapter = HTTPAdapter(max_retries=retry, pool_connections=max_workers, pool_maxsize=max_workers)
        self.session.mount("https://", adapter)
 
    def get_related(self, rxcui: str, ttys: List[str]) -> List[Dict]:
        self.limiter.wait()
        response = self.session.get(
            f"{self.BASE_URL}/rxcui/{rxcui}/related.json",
            params={"tty": " ".join(ttys)},
            timeout=REQUEST_TIMEOUT,
        )
        response.raise_for_status()
 
        groups = response.json().get("relatedGroup", {}).get("conceptGroup") or []
        results = []
        for group in groups:
            if group.get("tty") not in ttys:
                continue
            for prop in group.get("conceptProperties", []):
                results.append({
                    "rx_cui": prop.get("rxcui"),
                    "name": prop.get("name"),
                    "tty": prop.get("tty"),
                })
        return results
 
 
def process_brand(fetcher: ConcurrentRxNormFetcher, brand: Dict) -> Tuple[List[Dict], List[Dict]]:
    """Fetch all SBD products of one brand and their IN/PIN composition.
 
    Returns (rows, errors). A product whose composition fetch fails is NOT written
    with empty ingredients; it goes to `errors` instead so it can be retried.
    """
    brand_rxcui = brand.get("rx_cui")
    brand_name = brand.get("name")
    rows, errors = [], []
 
    try:
        products = fetcher.get_related(brand_rxcui, ["SBD"])
    except Exception as exc:
        errors.append({"rx_cui": brand_rxcui, "name": brand_name,
                       "product_rxcui": None, "product_name": None, "error": repr(exc)})
        return rows, errors
 
    for product in products:
        try:
            compositions = fetcher.get_related(product["rx_cui"], ["IN", "PIN"])
        except Exception as exc:
            errors.append({"rx_cui": brand_rxcui, "name": brand_name,
                           "product_rxcui": product.get("rx_cui"),
                           "product_name": product.get("name"), "error": repr(exc)})
            continue
 
        rows.append({
            "rx_cui": brand_rxcui,
            "name": brand_name,
            "product_rxcui": product.get("rx_cui"),
            "product_name": product.get("name"),
            "ingredients": [c for c in compositions if c.get("tty") == "IN"],
            "precise_ingredients": [c for c in compositions if c.get("tty") == "PIN"],
        })
    return rows, errors
```

```python
import pandas as pd

fetcher = RxNormConceptFetcher()
brands = fetcher.get_all_brand()
print(f"Total fetched brands: {len(brands)}") 

df_brands = pd.DataFrame(brands)
df_brands.to_csv(RXNORM_BRAND_PATH, index=False)
print(f"Brand RxNorm stored in {RXNORM_BRAND_PATH}")
```

```python
def fetch_all_brand_products(brands: List[Dict], max_workers: int = MAX_WORKERS,
                             rps: float = MAX_REQUESTS_PER_SECOND):
    fetcher = ConcurrentRxNormFetcher(max_workers=max_workers, rps=rps)
    results = [None] * len(brands)
 
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = {executor.submit(process_brand, fetcher, brand): idx
                   for idx, brand in enumerate(brands)}
        for future in tqdm(as_completed(futures), total=len(futures),
                           desc="Processing fetch products..."):
            results[futures[future]] = future.result()  # process_brand never raises
 
    # Flatten in the original brand order -> deterministic CSV regardless of thread timing
    all_rows = [row for rows, _ in results for row in rows]
    all_errors = [err for _, errs in results for err in errs]
    return all_rows, all_errors
 
 
brand_ingredients, inconsistent_ing = fetch_all_brand_products(brands)
    
```

```python
import pandas as pd

df_result = pd.DataFrame(brand_ingredients)
df_result.to_csv(RXNORM_BRAND_PRODUCT_PATH, index=False)
print(f"Brand Products RxNorm stored in {RXNORM_BRAND_PRODUCT_PATH}")

df_err = pd.DataFrame(inconsistent_ing)
df_err.to_csv(RXNORM_BRAND_ERR_PATH, index=False)
print(f"Brand RxNorm Err stored in {RXNORM_BRAND_ERR_PATH}")
```
