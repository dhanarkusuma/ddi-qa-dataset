### DDinter 2.0
DDinter 2.0 is a comprehensive, professional, and open-access database that provides extensive information on drug-drug interactions (DDIs), drug-food interactions (DFIs), drug-disease similarity interactions (DDSIs), and therapeutic duplications. This enhanced version offers expanded drug interaction data, along with detailed annotations such as mechanism descriptions, risk levels, management strategies, and updated ATC codes for both new and existing drugs. Curated by pharmacists, DDinter 2.0 serves as an invaluable resource for healthcare professionals, researchers, and patients, supporting clinical decision-making and promoting patient safety.

URL: https://ddinter2.scbdd.com/
DOI: https://doi.org/10.1093/nar/gkae726

```python
DATASET_DIR = "../../datasets/"
DATASET_REPOSITORY = "ddinter2.0"


import pathlib
OUTPUT_DIR = pathlib.Path(DATASET_DIR)
print(f"Dataset Directory: {OUTPUT_DIR}")

FILES_DDI = [
    "ddinter_downloads_code_A", 
    "ddinter_downloads_code_B",
    "ddinter_downloads_code_D",
    "ddinter_downloads_code_H",
    "ddinter_downloads_code_L",
    "ddinter_downloads_code_P",
    "ddinter_downloads_code_R",
    "ddinter_downloads_code_V"
]
DDI_CODES = {
    "A": "Interaction involving alimentary tract and metabolism drugs",
    "B": "Interaction involving blood and blood forming organs drugs",
    "D": "Interaction involving dermatologicals drugs",
    "H": "Interaction involving systemic hormonal preparations, excluding sex hormones and insulins drugs",
    "L": "Interaction involving antineoplastic and immunomodulating agents drugs",
    "P": "Interaction involving antiparasitic products, insecticides and repellents drugs",
    "R": "Interaction involving respiratory system drugs",
    "V": "Interaction involving various drugs"
}
DATASET_PATH = OUTPUT_DIR.joinpath(DATASET_REPOSITORY)
print(f"Drug-Drug Interactions Dataset Location: {DATASET_PATH}")
```

## Read Drug Drug Interactions Dataset

```python
import pandas as pd

datasets = {}

for FILE_DDI in FILES_DDI:
    df = pd.read_csv(f"{DATASET_PATH.joinpath(FILE_DDI)}.csv")
    datasets[FILE_DDI[len(FILE_DDI)-1:]] = df
```

```python
print("Category\tTotal Rows\tDescription")
total = 0
for k, v in datasets.items():
    print(f"Category {k}\t{v.shape[0]}\t\t{DDI_CODES[k]}")
    total += v.shape[0]

print(f"\nTotal DDI Rows: {total}")
```
